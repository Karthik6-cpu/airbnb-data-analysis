```python
# ==========================================
# AIRBNB DATA ANALYSIS AND PRICE PREDICTION
# PART 1: IMPORT AND LOAD DATA
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the Excel dataset
df = pd.read_excel("Airbnb_Open_Data.xlsx")

# Display basic information
print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Display first 5 rows
df.head()
```

```python
# ==========================================
# PART 2: UNDERSTAND AND CHECK THE DATASET
# ==========================================

# 1. Display all column names
print("COLUMN NAMES:")
print(df.columns.tolist())

# 2. Check data types and non-null values
print("\nDATASET INFORMATION:")
df.info()

# 3. Check missing values
print("\nMISSING VALUES:")
print(df.isnull().sum())

# 4. Check duplicate rows
print("\nNUMBER OF DUPLICATE ROWS:")
print(df.duplicated().sum())

# 5. Statistical summary
print("\nSTATISTICAL SUMMARY:")
df.describe(include="all")
```

```python
# ==========================================
# PART 3: DATA CLEANING
# ==========================================

# Make a copy of the original dataset
df_clean = df.copy()

# ------------------------------------------
# 1. Clean column names
# ------------------------------------------
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Cleaned column names:")
print(df_clean.columns.tolist())


# ------------------------------------------
# 2. Remove duplicate rows
# ------------------------------------------
duplicates = df_clean.duplicated().sum()

print("\nDuplicate rows found:", duplicates)

df_clean = df_clean.drop_duplicates()

print("Rows after removing duplicates:", len(df_clean))


# ------------------------------------------
# 3. Convert price to numeric
# ------------------------------------------
df_clean["price"] = (
    df_clean["price"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

df_clean["price"] = pd.to_numeric(
    df_clean["price"],
    errors="coerce"
)


# ------------------------------------------
# 4. Convert important numeric columns
# ------------------------------------------
numeric_columns = [
    "minimum_nights",
    "number_of_reviews",
    "reviews_per_month",
    "availability_365",
    "calculated_host_listings_count",
    "review_rate_number",
    "service_fee",
    "construction_year"
]

for column in numeric_columns:
    if column in df_clean.columns:
        df_clean[column] = pd.to_numeric(
            df_clean[column],
            errors="coerce"
        )


# ------------------------------------------
# 5. Remove rows where price is missing
# ------------------------------------------
df_clean = df_clean.dropna(subset=["price"])


# ------------------------------------------
# 6. Remove invalid prices
# ------------------------------------------
df_clean = df_clean[df_clean["price"] > 0]


# ------------------------------------------
# 7. Fill missing numerical values
# ------------------------------------------
numeric_columns = df_clean.select_dtypes(
    include=np.number
).columns

for column in numeric_columns:
    df_clean[column] = df_clean[column].fillna(
        df_clean[column].median()
    )


# ------------------------------------------
# 8. Fill missing categorical values
# ------------------------------------------
categorical_columns = df_clean.select_dtypes(
    include="object"
).columns

for column in categorical_columns:
    df_clean[column] = df_clean[column].fillna(
        "Unknown"
    )


# ------------------------------------------
# 9. Final dataset information
# ------------------------------------------
print("\n================================")
print("DATA CLEANING COMPLETED")
print("================================")

print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])

print("\nRemaining missing values:")
print(df_clean.isnull().sum().sum())

print("\nFirst 5 cleaned records:")
display(df_clean.head())
```

```python
# ==========================================
# PART 4: BASIC DATA ANALYSIS
# ==========================================

print("========== AIRBNB DATASET OVERVIEW ==========")

# Total number of listings
print("\n1. Total Number of Listings:")
print(df_clean.shape[0])

# Total number of columns
print("\n2. Total Number of Columns:")
print(df_clean.shape[1])


# ------------------------------------------
# 1. Average, minimum and maximum price
# ------------------------------------------

print("\n========== PRICE ANALYSIS ==========")

print("Average Price:", round(df_clean["price"].mean(), 2))
print("Minimum Price:", df_clean["price"].min())
print("Maximum Price:", df_clean["price"].max())
print("Median Price:", df_clean["price"].median())


# ------------------------------------------
# 2. Room Type Analysis
# ------------------------------------------

print("\n========== ROOM TYPE ANALYSIS ==========")

room_type_counts = df_clean["room_type"].value_counts()

print(room_type_counts)


# ------------------------------------------
# 3. Neighbourhood Analysis
# ------------------------------------------

print("\n========== TOP 10 NEIGHBOURHOODS ==========")

top_neighbourhoods = (
    df_clean["neighbourhood"]
    .value_counts()
    .head(10)
)

print(top_neighbourhoods)


# ------------------------------------------
# 4. Average Price by Room Type
# ------------------------------------------

print("\n========== AVERAGE PRICE BY ROOM TYPE ==========")

average_price_room = (
    df_clean
    .groupby("room_type")["price"]
    .mean()
    .sort_values(ascending=False)
)

print(average_price_room.round(2))


# ------------------------------------------
# 5. Average Price by Neighbourhood
# ------------------------------------------

print("\n========== TOP 10 EXPENSIVE NEIGHBOURHOODS ==========")

average_price_neighbourhood = (
    df_clean
    .groupby("neighbourhood")["price"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

print(average_price_neighbourhood.round(2))


# ------------------------------------------
# 6. Most Reviewed Listings
# ------------------------------------------

print("\n========== MOST REVIEWED LISTINGS ==========")

most_reviewed = (
    df_clean[
        ["name", "neighbourhood", "room_type", "price",
         "number_of_reviews"]
    ]
    .sort_values(
        "number_of_reviews",
        ascending=False
    )
    .head(10)
)

display(most_reviewed)


# ------------------------------------------
# 7. Average Availability by Room Type
# ------------------------------------------

print("\n========== AVAILABILITY ANALYSIS ==========")

average_availability = (
    df_clean
    .groupby("room_type")["availability_365"]
    .mean()
    .sort_values(ascending=False)
)

print(average_availability.round(2))


# ------------------------------------------
# 8. Final Summary
# ------------------------------------------

print("\n==========================================")
print("BASIC ANALYSIS COMPLETED")
print("==========================================")
```

```python
# ==========================================
# PART 8: MACHINE LEARNING
# AIRBNB PRICE PREDICTION
# ==========================================

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# ------------------------------------------
# 1. Select useful features
# ------------------------------------------

features = [
    "neighbourhood",
    "room_type",
    "minimum_nights",
    "number_of_reviews",
    "reviews_per_month",
    "availability_365"
]

target = "price"

# Keep only columns that actually exist
features = [col for col in features if col in df_clean.columns]

print("Features used:")
print(features)


# ------------------------------------------
# 2. Create model dataset
# ------------------------------------------

model_df = df_clean[features + [target]].copy()

# Remove missing target values
model_df = model_df.dropna(subset=[target])


# ------------------------------------------
# 3. Take a sample for faster training
# ------------------------------------------

sample_size = min(30000, len(model_df))

model_df = model_df.sample(
    n=sample_size,
    random_state=42
)

print("\nRecords used for machine learning:", len(model_df))


# ------------------------------------------
# 4. Separate X and y
# ------------------------------------------

X = model_df[features]
y = model_df[target]


# ------------------------------------------
# 5. Identify categorical/numerical columns
# ------------------------------------------

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_features = X.select_dtypes(
    include=[np.number]
).columns.tolist()

print("\nCategorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numerical_features)


# ------------------------------------------
# 6. Handle missing values
# ------------------------------------------

for column in numerical_features:
    X[column] = X[column].fillna(
        X[column].median()
    )

for column in categorical_features:
    X[column] = X[column].fillna(
        "Unknown"
    )


# ------------------------------------------
# 7. Split data
# ------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))


# ------------------------------------------
# 8. Preprocessing
# ------------------------------------------

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        )
    ],
    remainder="passthrough"
)


# ------------------------------------------
# 9. Create Random Forest model
# ------------------------------------------

model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),

        ("regressor",
         RandomForestRegressor(
             n_estimators=50,
             max_depth=15,
             random_state=42,
             n_jobs=-1
         )
        )
    ]
)


# ------------------------------------------
# 10. Train model
# ------------------------------------------

print("\nTraining the model...")

model.fit(
    X_train,
    y_train
)

print("Model training completed!")


# ------------------------------------------
# 11. Make predictions
# ------------------------------------------

y_pred = model.predict(X_test)


# ------------------------------------------
# 12. Evaluate model
# ------------------------------------------

mae = mean_absolute_error(
    y_test,
    y_pred
)

rmse = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred
    )
)

r2 = r2_score(
    y_test,
    y_pred
)


# ------------------------------------------
# 13. Display results
# ------------------------------------------

print("\n==========================================")
print("MODEL PERFORMANCE")
print("==========================================")

print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R² Score: {r2:.4f}")

print("==========================================")
```

```python
# ==========================================
# PART 9: ACTUAL VS PREDICTED PRICES
# ==========================================

import matplotlib.pyplot as plt
import numpy as np

# ------------------------------------------
# 1. Create Actual vs Predicted DataFrame
# ------------------------------------------

prediction_results = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred
})

print("Sample predictions:")
display(prediction_results.head(10))


# ------------------------------------------
# 2. Actual vs Predicted Graph
# ------------------------------------------

plt.figure(figsize=(9, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.4
)

# Reference line
minimum = min(y_test.min(), y_pred.min())
maximum = max(y_test.max(), y_pred.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum]
)

plt.title("Actual vs Predicted Airbnb Prices")
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")

plt.tight_layout()
plt.show()


# ------------------------------------------
# 3. Display Model Performance
# ------------------------------------------

print("==========================================")
print("FINAL MODEL RESULTS")
print("==========================================")

print(f"MAE  : {mae:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"R²   : {r2:.4f}")

print("==========================================")
```
